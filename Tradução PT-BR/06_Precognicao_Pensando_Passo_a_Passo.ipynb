## Lição

Se alguém te acordasse e imediatamente começasse a fazer várias perguntas complicadas às quais você tivesse que responder na hora, como você se sairia? Provavelmente não tão bem quanto se tivesse tempo para **pensar na resposta primeiro**.

Adivinha? Claude funciona da mesma forma.

**Dar tempo ao Claude para pensar passo a passo às vezes o torna mais preciso**, especialmente em tarefas complexas. No entanto, **pensar só conta quando é feito em voz alta**. Você não pode pedir ao Claude que pense mas produza apenas a resposta — nesse caso, nenhum raciocínio terá ocorrido de fato.

### Exemplos

No prompt abaixo, é claro para um leitor humano que a segunda frase contradiz a primeira. Mas **Claude interpreta a palavra "unrelated" de forma demasiadamente literal**.

```python
# Prompt
PROMPT = """Is this movie review sentiment positive or negative?

This movie blew my mind with its freshness and originality. In totally unrelated news, I have been living under a rock since the year 1900."""

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

Para melhorar a resposta do Claude, vamos **permitir que ele raciocine antes de responder**. Fazemos isso detalhando explicitamente as etapas que o Claude deve seguir para processar e pensar sobre sua tarefa. Combinado com um toque de role prompting, isso permite que o Claude compreenda a resenha de forma mais profunda.

```python
# System prompt
SYSTEM_PROMPT = "You are a savvy reader of movie reviews."

# Prompt
PROMPT = """Is this review sentiment positive or negative? First, write the best arguments for each side in <positive-argument> and <negative-argument> XML tags, then answer.

This movie blew my mind with its freshness and originality. In totally unrelated news, I have been living under a rock since 1900."""

# Imprime a resposta do Claude
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

**Claude às vezes é sensível à ordem**. Este exemplo está no limite da capacidade do Claude de compreender textos com nuances, e quando invertemos a ordem dos argumentos do exemplo anterior — colocando o negativo primeiro e o positivo segundo — isso muda a avaliação geral do Claude para positiva.

Na maioria das situações (mas não em todas, o que pode ser confuso), **Claude tende a escolher a segunda de duas opções**, possivelmente porque nos dados de treinamento da web, a segunda opção era mais frequentemente a correta.

```python
# Prompt
PROMPT = """Is this review sentiment negative or positive? First write the best arguments for each side in <negative-argument> and <positive-argument> XML tags, then answer.

This movie blew my mind with its freshness and originality. Unrelatedly, I have been living under a rock since 1900."""

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

**Deixar o Claude pensar pode mudar sua resposta de incorreta para correta**. Em muitos casos em que o Claude erra, é simples assim!

Vamos analisar um exemplo onde a resposta do Claude está incorreta para ver como pedir que ele pense pode corrigir isso.

```python
# Prompt
PROMPT = "Name a famous movie starring an actor who was born in the year 1956."

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```
