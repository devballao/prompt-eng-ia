## Lição

**O Claude responde melhor a instruções claras e diretas.**

Pense no Claude como qualquer outro ser humano que é novo no trabalho. **O Claude não tem contexto** sobre o que fazer além do que você literalmente diz a ele. Assim como quando você instrui um humano pela primeira vez em uma tarefa, quanto mais você explica exatamente o que quer de forma direta ao Claude, melhor e mais precisa será a resposta dele."				
				
Em caso de dúvida, siga a **Regra de Ouro do Prompt Claro**:
- Mostre seu prompt a um colega ou amigo e peça que ele mesmo siga as instruções para ver se consegue produzir o resultado desejado. Se ele ficar confuso, o Claude também ficará.				

### Exemplos

Vamos pegar uma tarefa como escrever poesia. (Ignore qualquer incompatibilidade de sílabas — LLMs ainda não são bons em contar sílabas.)

```python
# Prompt
PROMPT = "Write a haiku about robots."

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

Este haiku é bem agradável, mas os usuários podem querer que o Claude vá direto ao poema sem o preâmbulo "Here is a haiku".

Como conseguimos isso? **Pedimos diretamente**!

```python
# Prompt
PROMPT = "Write a haiku about robots. Skip the preamble; go straight into the poem."

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

Aqui está outro exemplo. Vamos perguntar ao Claude quem é o melhor jogador de basquete de todos os tempos. Você pode ver abaixo que, embora o Claude liste alguns nomes, **ele não responde com um "melhor" definitivo**.

```python
# Prompt
PROMPT = "Who is the best basketball player of all time?"

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

Podemos fazer o Claude tomar uma decisão e escolher o melhor jogador? Sim! Basta pedir!

```python
# Prompt
PROMPT = "Who is the best basketball player of all time? Yes, there are differing opinions, but if you absolutely had to pick one player, who would it be?"

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

### Exercício 2.1 - Espanhol
Modifique o `SYSTEM_PROMPT` para fazer o Claude responder em espanhol.

```python
# System prompt - este é o único campo que você deve alterar
SYSTEM_PROMPT = "[Replace this text]"

# Prompt
PROMPT = "Hello Claude, how are you?"

# Obtém a resposta do Claude
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Função para avaliar a correção do exercício
def grade_exercise(text):
    return "hola" in text.lower()

# Imprime a resposta do Claude e a avaliação correspondente
print(response)
print("\n--------------------------- AVALIAÇÃO ---------------------------")
print("Este exercício foi resolvido corretamente:", grade_exercise(response))
```

### Exercício 2.3 - Escreva uma História

Modifique o `PROMPT` para que o Claude responda com a resposta mais longa que você conseguir. Se sua resposta tiver **mais de 800 palavras**, a resposta do Claude será avaliada como correta.

```python
# Prompt - este é o único campo que você deve alterar
PROMPT = "[Replace this text]"

# Obtém a resposta do Claude
response = get_completion(PROMPT)

# Função para avaliar a correção do exercício
def grade_exercise(text):
    trimmed = text.strip()
    words = len(trimmed.split())
    return words >= 800

# Imprime a resposta do Claude e a avaliação correspondente
print(response)
print("\n--------------------------- AVALIAÇÃO ---------------------------")
print("Este exercício foi resolvido corretamente:", grade_exercise(response))
```
