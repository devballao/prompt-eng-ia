## Lição

Continuando com o tema de que o Claude não tem contexto além do que você diz, às vezes é importante **instruir o Claude a assumir um papel específico (incluindo todo o contexto necessário)**. Isso também é conhecido como role prompting. Quanto mais detalhado o contexto do papel, melhor.

**Preparar o Claude com um papel pode melhorar seu desempenho** em diversas áreas, desde escrita até programação e resumo. É como quando os humanos às vezes são ajudados quando recebem a instrução de "pensar como um ______". O role prompting também pode alterar o estilo, o tom e a forma das respostas do Claude.

**Observação:** O role prompting pode ocorrer tanto no system prompt quanto como parte da mensagem do usuário.

### Exemplos

No exemplo abaixo, vemos que sem role prompting, o Claude fornece uma **resposta direta e sem estilização** quando solicitado a dar uma perspectiva em uma única frase sobre skate.

No entanto, quando preparamos o Claude para assumir o papel de um gato, a perspectiva do Claude muda e, assim, **o tom, o estilo e o conteúdo da resposta se adaptam ao novo papel**.

**Observação:** Uma técnica bônus que você pode usar é **fornecer ao Claude contexto sobre seu público-alvo**. Abaixo, poderíamos ter ajustado o prompt para também dizer ao Claude com quem ele deveria estar falando. "You are a cat" produz uma resposta bem diferente de "you are a cat talking to a crowd of skateboarders".

Aqui está o prompt sem role prompting no system prompt:

```python
# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

Aqui está a mesma pergunta do usuário, mas com role prompting.

```python
# System prompt
SYSTEM_PROMPT = "You are a cat."

# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Imprime a resposta do Claude
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

Você pode usar o role prompting como forma de fazer o Claude emular certos estilos de escrita, falar em uma determinada voz ou guiar a complexidade das suas respostas. **O role prompting também pode fazer o Claude se sair melhor em tarefas de matemática ou lógica.**

Por exemplo, no exemplo abaixo, há uma resposta correta definitiva, que é sim. No entanto, o Claude erra e acha que faltam informações, o que não é verdade:

```python
# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don't know if Anne is married. Is a married person looking at an unmarried person?"

# Imprime a resposta do Claude
print(get_completion(PROMPT))
```

Agora, e se **prepararmos o Claude para agir como um bot de lógica**? Como isso mudará a resposta do Claude?

Acontece que com essa nova atribuição de papel, o Claude acerta. (Embora notavelmente nem sempre pelos motivos corretos)

```python
# System prompt
SYSTEM_PROMPT = "You are a logic bot designed to answer complex logic problems."

# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don't know if Anne is married. Is a married person looking at an unmarried person?"

# Imprime a resposta do Claude
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

**Observação:** O que você aprenderá ao longo deste curso é que existem **muitas técnicas de engenharia de prompt que você pode usar para obter resultados semelhantes**. Quais técnicas usar depende de você e da sua preferência! Encorajamos você a **experimentar para encontrar seu próprio estilo de engenharia de prompt**.

### Exercício 3.1 - Correção Matemática
Em alguns casos, **o Claude pode ter dificuldades com matemática**, mesmo com matemática simples. Abaixo, o Claude avalia incorretamente o problema matemático como resolvido corretamente, mesmo havendo um erro aritmético óbvio no segundo passo. Observe que o Claude realmente detecta o erro ao percorrer passo a passo, mas não chega à conclusão de que a solução geral está errada.

Modifique o `PROMPT` e/ou o `SYSTEM_PROMPT` para fazer o Claude avaliar a solução como `incorrectly` resolvida, em vez de corretamente resolvida.


```python
# System prompt - se não quiser usar um system prompt, você pode deixar esta variável como uma string vazia
SYSTEM_PROMPT = ""

# Prompt
PROMPT = """Is this equation solved correctly below?

2x - 3 = 9
2x = 6
x = 3"""

# Obtém a resposta do Claude
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Função para avaliar a correção do exercício
def grade_exercise(text):
    if "incorrect" in text or "not correct" in text.lower():
        return True
    else:
        return False

# Imprime a resposta do Claude e a avaliação correspondente
print(response)
print("\n--------------------------- AVALIAÇÃO ---------------------------")
print("Este exercício foi resolvido corretamente:", grade_exercise(response))
```
