## Lição

**O Claude pode formatar sua saída de diversas maneiras**. Você só precisa pedir que ele faça isso!

Uma dessas formas é usando tags XML para separar a resposta de qualquer outro texto supérfluo. Você já aprendeu que pode usar tags XML para tornar seu prompt mais claro e fácil de ser interpretado pelo Claude. Acontece que você também pode pedir ao Claude para **usar tags XML para tornar sua saída mais clara e facilmente compreensível** para humanos.

### Exemplos

Lembra do "problema do preâmbulo do poema" que resolvemos no Capítulo 2 pedindo ao Claude para pular o preâmbulo completamente? Acontece que podemos obter um resultado semelhante **dizendo ao Claude para colocar o poema em tags XML**.

```python
# Conteúdo variável
ANIMAL = "Rabbit"

# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Please write a haiku about {ANIMAL}. Put it in <haiku> tags."

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print(PROMPT)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(get_completion(PROMPT))
```

Por que faríamos isso? Bem, ter a saída em **tags XML permite que o usuário final obtenha de forma confiável apenas o poema, escrevendo um pequeno programa para extrair o conteúdo entre as tags XML**.

Uma extensão dessa técnica é **colocar a primeira tag XML no turno do `assistant`. Quando você coloca texto no turno do `assistant`, está basicamente dizendo ao Claude que ele já disse algo, e que deve continuar a partir daquele ponto. Essa técnica é chamada de "falar pelo Claude" ou "preencher previamente a resposta do Claude" (prefilling).

Abaixo, fizemos isso com a primeira tag XML `<haiku>`. Observe como o Claude continua diretamente de onde paramos.

```python
# Conteúdo variável
ANIMAL = "Cat"

# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Please write a haiku about {ANIMAL}. Put it in <haiku> tags."

# Prefill para a resposta do Claude
PREFILL = "<haiku>"

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print("TURNO DO USUÁRIO:")
print(PROMPT)
print("\nTURNO DO ASSISTENTE:")
print(PREFILL)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(get_completion(PROMPT, prefill=PREFILL))
```

O Claude também se destaca no uso de outros estilos de formatação de saída, notavelmente `JSON`. Se você quiser impor a saída em JSON (não de forma determinística, mas próximo disso), também pode preencher previamente a resposta do Claude com o colchete de abertura, `{`}.

```python
# Conteúdo variável
ANIMAL = "Cat"

# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Please write a haiku about {ANIMAL}. Use JSON format with the keys as \"first_line\", \"second_line\", and \"third_line\"."

# Prefill para a resposta do Claude
PREFILL = "{"

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print("TURNO DO USUÁRIO")
print(PROMPT)
print("\nTURNO DO ASSISTENTE")
print(PREFILL)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(get_completion(PROMPT, prefill=PREFILL))
```

Abaixo está um exemplo de **múltiplas variáveis de entrada no mesmo prompt E especificação de formatação de saída, tudo feito usando tags XML**.

```python
# Primeira variável de entrada
EMAIL = "Hi Zack, just pinging you for a quick update on that prompt you were supposed to write."

# Segunda variável de entrada
ADJECTIVE = "olde english"

# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Hey Claude. Here is an email: <email>{EMAIL}</email>. Make this email more {ADJECTIVE}. Write the new version in <{ADJECTIVE}_email> XML tags."

# Prefill para a resposta do Claude (agora como uma f-string com uma variável)
PREFILL = f"<{ADJECTIVE}_email>"

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print("TURNO DO USUÁRIO")
print(PROMPT)
print("\nTURNO DO ASSISTENTE")
print(PREFILL)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(get_completion(PROMPT, prefill=PREFILL))
```

#### Lição bônus

Se você está chamando o Claude pela API, pode passar a tag XML de fechamento para o parâmetro `stop_sequences` para que o Claude pare de gerar assim que emitir a tag desejada. Isso pode economizar dinheiro e tempo até o último token, eliminando os comentários finais do Claude depois que ele já forneceu a resposta que você precisava.

### Exercício 5.1 - Steph Curry GOAT
Forçado a fazer uma escolha, o Claude designa Michael Jordan como o melhor jogador de basquete de todos os tempos. Conseguimos fazer o Claude escolher outra pessoa?

Altere a variável `PREFILL` para **compelir o Claude a fazer um argumento detalhado de que o melhor jogador de basquete de todos os tempos é Stephen Curry**. Tente não alterar nada além do `PREFILL`, pois esse é o foco deste exercício.

```python
# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Who is the best basketball player of all time? Please choose one specific player."

# Prefill para a resposta do Claude
PREFILL = ""

# Obtém a resposta do Claude
response = get_completion(PROMPT, prefill=PREFILL)

# Função para avaliar a correção do exercício
def grade_exercise(text):
    return bool(re.search("Warrior", text))

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print("TURNO DO USUÁRIO")
print(PROMPT)
print("\nTURNO DO ASSISTENTE")
print(PREFILL)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(response)
print("\n------------------------------------------ AVALIAÇÃO ------------------------------------------")
print("Este exercício foi resolvido corretamente:", grade_exercise(response))
```

### Exercício 5.2 - Dois Haikus
Modifique o `PROMPT` abaixo usando tags XML para que o Claude escreva dois haikus sobre o animal em vez de apenas um. Deve ficar claro onde um poema termina e o outro começa.

```python
# Conteúdo variável
ANIMAL = "cats"

# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Please write a haiku about {ANIMAL}. Put it in <haiku> tags."

# Prefill para a resposta do Claude
PREFILL = "<haiku>"

# Obtém a resposta do Claude
response = get_completion(PROMPT, prefill=PREFILL)

# Função para avaliar a correção do exercício
def grade_exercise(text):
    return bool(
        (re.search("cat", text.lower()) and re.search("<haiku>", text))
        and (text.count("\n") + 1) > 5
    )

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print("TURNO DO USUÁRIO")
print(PROMPT)
print("\nTURNO DO ASSISTENTE")
print(PREFILL)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(response)
print("\n------------------------------------------ AVALIAÇÃO ------------------------------------------")
print("Este exercício foi resolvido corretamente:", grade_exercise(response))
```

### Exercício 5.3 - Dois Haikus, Dois Animais
Modifique o `PROMPT` abaixo para que **o Claude produza dois haikus sobre dois animais diferentes**. Use `{ANIMAL1}` como marcador para a primeira substituição e `{ANIMAL2}` como marcador para a segunda substituição.

```python
# Primeira variável de entrada
ANIMAL1 = "Cat"

# Segunda variável de entrada
ANIMAL2 = "Dog"

# Template de prompt com um marcador para o conteúdo variável
PROMPT = f"Please write a haiku about {ANIMAL1}. Put it in <haiku> tags."

# Obtém a resposta do Claude
response = get_completion(PROMPT)

# Função para avaliar a correção do exercício
def grade_exercise(text):
    return bool(re.search("tail", text.lower()) and re.search("cat", text.lower()) and re.search("<haiku>", text))

# Imprime a resposta do Claude
print("--------------------------- Prompt completo com substituições de variáveis ---------------------------")
print("TURNO DO USUÁRIO")
print(PROMPT)
print("\n------------------------------------- Resposta do Claude -------------------------------------")
print(response)
print("\n------------------------------------------ AVALIAÇÃO ------------------------------------------")
print("Este exercício foi resolvido corretamente:", grade_exercise(response))
```
